```python
import COG_Conversion


EXTENSIONS_TO_CONVERT = ['tif'] #! Special drivers may be required for certain file extensions

INPUT_FOLDER = r'W:\ilmb\vic\geobc\Workarea\mgraff\objectstorage\imagefiles\oldext\ElephantHill_Imagery'
OUTPUT_FOLDER = r'W:\ilmb\vic\geobc\Workarea\mgraff\objectstorage\imagefiles\oldext\ElephantHill_Imagery\COG' 
```

```python
# cog conversion of folder
COG_Conversion.find_and_convert_files(INPUT_FOLDER, OUTPUT_FOLDER, EXTENSIONS_TO_CONVERT)
```

```python
import S3_Uploader
import os

OBJECT_NAME_CUTOFF = r'W:/ilmb/vic/geobc/Workarea/'

UPLOAD_FOLDER = OUTPUT_FOLDER
BUCKET_NAME = 'cloudgistest'

# Create the dictionary
UPLOAD_DICT = {}
for file_path in OUTPUT_FOLDER:
    # Compute the object path by removing the cutoff
    if file_path.startswith(OBJECT_NAME_CUTOFF):
        object_path = file_path[len(OBJECT_NAME_CUTOFF):].lstrip(os.sep)
        # Add to the dictionary
        UPLOAD_DICT[file_path] = object_path



# s3 upload of folder
links = S3_Uploader.multi_upload(UPLOAD_DICT, BUCKET_NAME)

for link in links:
    print(link)
```
