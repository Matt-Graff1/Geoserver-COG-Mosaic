## Issues with STAC-API Mosaic

**Server limits**:
When mosaicing large areas, the EarthSearch AWS server hosting the granules returned a ```HTTP error code 502```. \
This was fixed by limiting the features to 10 (hardLimit parameter on the STAC data store in GeoServer) \
Something that we should think about is the amount of requests the server can handle - it will be a lot if indexing mosaics
