# Tutorial: Installing GeoServer and the COG Plugin on Windows

This guide will walk you through installing GeoServer and the Cloud Optimized GeoTIFF (COG) plugin on a Windows system.

## 1. Download and Install GeoServer

- **Download GeoServer**:
  - Visit the official GeoServer download page: [https://geoserver.org/download/](https://geoserver.org/download/).
  - Choose the Windows Installer version (`.exe`) and download it.

- **Run the Installer**:
  - Locate the downloaded `.exe` file and double-click it to start the installation.
  - **Choose Installation Directory**: Select the directory where you want to install GeoServer or leave it as the default.
  - **Set Installation Options**:
    - Choose the Java Runtime Environment (JRE) you wish to use if prompted.
    - Set the GeoServer port number (default is `8080`).
  - **Complete Installation**: Click through the remaining steps to complete the installation process.

- **Start GeoServer**:
  - After installation, you can start GeoServer either from the Start Menu or by running the `Start GeoServer` shortcut from your desktop.
  - Once started, you can access GeoServer at `http://localhost:8080/geoserver`.

## 2. Download and Install the COG Plugin

- **Download the COG Plugin**:
  - Visit the GeoServer plugin page: [https://geoserver.org/release/stable/](https://geoserver.org/release/stable/).
  - Find the COG plugin under "Community Modules" and download the ZIP file that matches your GeoServer version.

- **Install the Plugin**:
  - **Extract the Plugin Files**: Extract the contents of the ZIP file.
  - **Copy Plugin Files**: Copy the extracted `.jar` files to the `WEB-INF/lib` directory of your GeoServer installation.
    - Example path: `C:\Program Files (x86)\GeoServer\webapps\geoserver\WEB-INF\lib`
  - **Restart GeoServer**: After copying the files, restart GeoServer to load the new plugin.

## 3. Verify Plugin Installation

- **Access GeoServer Web Interface**:
  - Open your web browser and navigate to `http://localhost:8080/geoserver`.
  - **Log in** using your GeoServer credentials.
  - **Check the Data Stores**:
    - Go to "Stores" and try to create a new data store.
    - You should now see options related to Cloud Optimized GeoTIFF (COG) if the plugin is installed correctly.

## 4. Start Using GeoServer with COGs

- You can now create new layers in GeoServer using COGs as your data source.
- The COG format allows for efficient reading and streaming of large geospatial datasets directly from cloud storage or local storage.

---

By following these steps, you should have GeoServer and the COG plugin successfully installed on your Windows machine, allowing you to manage and serve Cloud Optimized GeoTIFFs through GeoServer.
